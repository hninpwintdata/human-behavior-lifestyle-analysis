```python
stress_by_occupation = df.groupby("Occupation")["Stress_Level"].mean()

stress_by_occupation.plot(kind="bar", figsize=(8, 5))

plt.title("Average Stress Level by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Stress Level")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
satisfaction_by_occupation = df.groupby("Occupation")["Satisfaction_Level"].mean()

satisfaction_by_occupation.plot(kind="bar", figsize=(8, 5))

plt.title("Average Satisfaction Level by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Satisfaction Level")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sleep_by_occupation = df.groupby("Occupation")["Sleep_Hours"].mean()

sleep_by_occupation.plot(kind="bar")

plt.title("Average Sleep Hours by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Sleep Hours")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
screen_time_by_occupation = df.groupby("Occupation")["Screen_Time_Hours"].mean()

screen_time_by_occupation.plot(kind="bar")

plt.title("Average Screen Time by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Screen Time (Hours)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
spending_by_occupation = df.groupby("Occupation")["Monthly_Spending"].mean()

spending_by_occupation.plot(kind="bar")

plt.title("Average Monthly Spending by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Monthly Spending (¥)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Key Findings

1. **Exercise and stress:** Participants who exercised more frequently generally reported lower average stress levels.
2. **Exercise and satisfaction:** Participants who exercised more frequently generally reported higher average satisfaction.
3. **Screen time and sleep:** Higher screen time was associated with fewer hours of sleep.
4. **Stress and satisfaction:** Higher stress levels were associated with lower satisfaction.
5. **Sleep and stress:** Longer sleep duration was associated with lower stress.
6. **Social media and stress:** More social media use showed a positive association with stress.

### Conclusion

The analysis identified several relationships between lifestyle habits and well-being in this dataset. Exercise frequency, sleep, screen time, and stress showed interesting patterns worth exploring further. These findings describe associations and do not prove cause and effect.

*Note: This dataset is used for practice and should not be treated as representative of the general population.*
```
